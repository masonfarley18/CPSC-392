```python
unfiltered = pd.DataFrame(penguins)

adelie_df = unfiltered[unfiltered['species'] == 'Adelie']
bl_Adelie = adelie_df["bill_length_mm"].sum() / len(adelie_df)

chinstrap_df = unfiltered[unfiltered['species'] == 'Chinstrap']
bl_Chinstrap = chinstrap_df["bill_length_mm"].sum() / len(chinstrap_df)

gentoo_df = unfiltered[unfiltered['species'] == 'Gentoo']
bl_Gentoo = gentoo_df["bill_length_mm"].sum() / len(gentoo_df)

print(bl_Adelie)
print(bl_Chinstrap)
print(bl_Gentoo)
```

```python
first_df = titanic[titanic['class'] == 'First']
first_prop_survivors = len(first_df[first_df['survived'] == 1]) / len(first_df)

second_df = titanic[titanic['class'] == 'Second']
second_prop_survivors = len(second_df[second_df['survived'] == 1]) / len(second_df)

third_df = titanic[titanic['class'] == 'Third']
third_prop_survivors = len(third_df[third_df['survived'] == 1]) / len(third_df)

print(first_prop_survivors)
print(second_prop_survivors)
print(third_prop_survivors)
```

```python
male_df = titanic[titanic['sex'] == 'male']
male_fare_mean = (male_df.loc[:, 'fare'].sum()) / len(male_df)

female_df = titanic[titanic['sex'] == 'female']
female_fare_mean = (female_df.loc[:, 'fare'].sum()) / len(female_df)

print(male_fare_mean)
print(female_fare_mean)
```
